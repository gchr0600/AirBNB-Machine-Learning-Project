This next section uses the amenities column to add an addition 155 dummy variables for different categories.

```python
import warnings
from pandas.errors import PerformanceWarning

# Suppress PerformanceWarnings from pandas
warnings.simplefilter(action='ignore', category=PerformanceWarning)

import ast
san_diego = raw_data.copy()

san_diego['amenities_test'] = san_diego['amenities'].apply(ast.literal_eval)


longest_row_ind = san_diego['amenities_test'].apply(len).idxmax()
print(longest_row_ind)

amenity_indices = list(range(12290, 12301))

longest_amenities_list = san_diego.loc[longest_row_ind, 'amenities_test']

for index in amenity_indices:
  list_of_amenities = san_diego.loc[index, 'amenities_test']
  for amenity in list_of_amenities:
        if amenity not in longest_amenities_list:
            longest_amenities_list.append(amenity)

for amenity in longest_amenities_list:
    san_diego[amenity] = san_diego['amenities'].str.contains(str(amenity)).astype(int)


for col in san_diego.columns:
  print(col)
```
